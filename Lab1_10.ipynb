```python
def sum_compare(n):
    """Возвращает (S1, S2, S2 - S1) для целого n >= 1."""
    if n < 1:
        raise ValueError("n должно быть не меньше 1")

    s1 = sum(i ** 2 for i in range(1, n + 1))
    s2 = sum(range(1, n + 1)) ** 2

    return s1, s2, s2 - s1


```

```python
# Два примера вызова функции sum_compare
result_4 = sum_compare(4)
result_10 = sum_compare(10)

print(f"sum_compare(4)  = {result_4}")
print(f"sum_compare(10) = {result_10}")

print(f"Для n=4:  S1={result_4[0]}, S2={result_4[1]}, S2-S1={result_4[2]}")
print(f"Для n=10: S1={result_10[0]}, S2={result_10[1]}, S2-S1={result_10[2]}")
```

```python
try:
    n = int(input("Введите целое число n ≥ 1: "))
    result = sum_compare(n)

    print(f"sum_compare({n}) → {result}")
    print(f"Разность S2 - S1 = {result[2]}")
except ValueError as error:
    print(f"Ошибка: {error}")
```
